To analyze a chirp we will need the chirp current that we injected into the cell. If you know how your recording program created the chirp then you can just recreate it. Otherwise if you used something like an ATF file you will need to load the file into python.
The reason we need the actual chirp stimulus is that we have to remove the frequency content of the chirp from the cell's voltage. The cell voltage will oscillate at the frequency the chirp is at no matter what but we want to see how the membrane reacts without the influence. We can do this through deconvolution in the frequency domain. Lastly we will need the start and stop frequency the chirp was run at. Deconvolution is a very noisy operation. Since the chirp is pure is does not contain frequencies above the frequency it is ended at. This causes the impedance to explode towards infinity with increasing frequency. So for the analysis we just avoid the frequencies below and above what we specified our chirp at.
